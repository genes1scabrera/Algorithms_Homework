1. O(1) means constant time and taking the same amount of time no matter
how much data there is. O(log N) gets to a solution faster by the cutting
a problem into smaller parts. O(N) checks on each item one time. Finally,
O(N^2) checks items using a loop inside a loop and tends to take much longer.

2. Binary search works on a sorted array by checking the middle number. If
the number is too small, it looks at the right half. If the number is too big,
it looks at the left half. It keeps doing that until it finds the number.

```python
def square_number(number):
    return number * number

print(square_number(4))
```
