Problem 1: Create a Node Class for a binary tree and create 8 node objects using the Node Class

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)
node5 = Node(5)
node6 = Node(6)
node7 = Node(7)
node8 = Node(8)
```

Problem 2: Create a Binary Tree, using the nodes created in #1, construct a binary tree with a depth of 3

```python
node1.left = node2
node1.right = node3
node2.left = node4
node2.right = node5
node3.left = node6
node3.right = node7
node4.left = node8
root = node1
```

Problem 3: Traverse the binary tree using each of the following methods
a. In-order traversal
b. Pre-order traversal
c. Post-order traversal

```python
# 3a
def inorder(node):
    if node is not None:
        inorder(node.left)
        print(node.value, end=" ")
        inorder(node.right)
# 3b
def preorder(node):
    if node is not None:
        print(node.value, end=" ")
        preorder(node.left)
        preorder(node.right)
# 3c
def postorder(node):
    if node is not None:
        postorder(node.left)
        postorder(node.right)
        print(node.value, end=" ")

print("In-order Traversal:")
inorder(root)
print("\nPre-order Traversal:")
preorder(root)
print("\nPost-order Traversal:")
postorder(root)
```
