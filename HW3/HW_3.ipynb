Problem 1 - Merge Two Sorted Arrays

```python
def merge_two_sorted_arrays(arr1, arr2):
    merged = []
    i = 0
    j = 0

    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1

    while i < len(arr1):
        merged.append(arr1[i])
        i += 1

    while j < len(arr2):
        merged.append(arr2[j])
        j += 1

    return merged


arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8]

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Merged array:", merge_two_sorted_arrays(arr1, arr2))
```

Problem 2: Iterative Merge Sort

```python
def merge_sort_iterative(arr):
    result = arr.copy()
    size = 1

    while size < len(result):
        for start in range(0, len(result), 2 * size):
            mid = min(start + size, len(result))
            end = min(start + 2 * size, len(result))

            left = result[start:mid]
            right = result[mid:end]
            merged = []
            i = 0
            j = 0

            while i < len(left) and j < len(right):
                if left[i] <= right[j]:
                    merged.append(left[i])
                    i += 1
                else:
                    merged.append(right[j])
                    j += 1

            while i < len(left):
                merged.append(left[i])
                i += 1

            while j < len(right):
                merged.append(right[j])
                j += 1

            result[start:end] = merged

        size *= 2

    return result

numbers = [8, 3, 5, 4, 7, 6, 1, 2]

print("Original array:", numbers)
print("Sorted array:", merge_sort_iterative(numbers))
```

Problem 3: Recursive Merge Sort

```python
def merge_sort_recursive(arr):
    if len(arr) <= 1:
        return arr.copy()

    mid = len(arr) // 2

    left = merge_sort_recursive(arr[:mid])
    right = merge_sort_recursive(arr[mid:])

    merged = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1

    while i < len(left):
        merged.append(left[i])
        i += 1

    while j < len(right):
        merged.append(right[j])
        j += 1

    return merged

numbers = [8, 3, 5, 4, 7, 6, 1, 2]

print("Original array:", numbers)
print("Sorted array:", merge_sort_recursive(numbers))
```
