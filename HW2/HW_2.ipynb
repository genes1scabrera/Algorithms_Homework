# Problem 1: Practice #1

Given an array of integers, nums, where each integer is between 1 and n
(n is the size of the array), some elements appear twice and others appear
once. Return an array of all the integers that appear twice in the array. You
must return the integers in their original order.

```python
def find_duplicates(nums):
    seen = set()
    duplicates = []

    for num in nums:
        if num in seen:
            duplicates.append(num)
        else:
            seen.add(num)

    return duplicates


nums = [4, 3, 2, 7, 8, 2, 3, 1]

print("Input:", nums)
print("Duplicates:", find_duplicates(nums))
```

# Problem 2: Merge Sort

Review Merge Sort. Write a Python function that implements Merge Sort and provide an example demonstrating how it works.

Merge Sort is a sorting algorithm that divides an array into smaller parts until each part contains only one element. It then merges those smaller parts back together in sorted order. The time complexity of Merge Sort is O(n log n).

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    return merge(left_half, right_half)


def merge(left, right):
    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result
```

```python
numbers = [8, 3, 5, 4, 7, 6, 1, 2]

sorted_numbers = merge_sort(numbers)

print("Original array:", numbers)
print("Sorted array:", sorted_numbers)
```

# Problem 3: Insertion Sort

Study Insertion Sort and provide an analysis of its time complexity.

Insertion Sort sorts an array by building a sorted section one element at a time. Each new element is compared with the elements before it, and larger elements are shifted to the right until the new element is placed in the correct position.

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):

        key = arr[i]

        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


numbers = [5, 2, 4, 6, 1, 3]

print("Original array:", numbers)

sorted_numbers = insertion_sort(numbers.copy())

print("Sorted array:", sorted_numbers)
```

# Time Complexity Analysis

The best case is O(n). The array is already sorted and each element only needs to be compared with the element before it.

The worst case is O(n^2), where elements may need to be compared with and shifted past several prevoius elements, and the array may be in reverse order.
